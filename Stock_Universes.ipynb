```python
# @title Import Dependencies
import pandas as pd
from finvizfinance.screener.overview import Overview

import warnings
warnings.filterwarnings('ignore')

def get_universe(filter : dict) -> pd.DataFrame:

  overview = Overview()
  overview.set_filter(filters_dict=filter)
  universe = overview.screener_view()
  # Remove the first character from the 'Ticker' column
  universe['Ticker'] = universe['Ticker'].str[1:]

  return universe
```

```python
# @title Mega Cap Stocks
'''
Universe: Mega Cap Stocks
Description: Mega-cap stocks are defined as having a market capitalization of $200 billion USD or higher.
'''

mega_cap = get_universe(filter={'Market Cap.': 'Mega ($200bln and more)'})
mega_cap = mega_cap[['Ticker', 'Company', 'Sector', 'Industry', 'Country']]
mega_cap.to_csv("mega_cap.csv")
print("Returned", mega_cap.shape[0],"companies")
print(mega_cap.tail().to_string())
```

## Combine All Market Cap DataFrames

```python
import pandas as pd

# List of all generated CSV files
csv_files = [
    'mega_cap.csv',
    'large_cap.csv',
    'mid_cap.csv',
    'small_cap.csv',
    'micro_cap.csv',
    'nano_cap.csv',
    'etfs.csv',
    'sp500.csv',
    'russell2000.csv',
    'nasdaq100.csv',
    'djia.csv',
    'all_tickers.csv'
]

# Initialize an empty list to store individual DataFrames
df_list = []

# Read each CSV file and append to the list
for file in csv_files:
    try:
        df = pd.read_csv(file, index_col=0) # Assuming the first column is an unnamed index
        df_list.append(df)
    except FileNotFoundError:
        print(f"Warning: File {file} not found. Skipping.")

# Concatenate all DataFrames into a single DataFrame
all_cap_df = pd.concat(df_list, ignore_index=True)

print(f"Combined DataFrame shape before dropping duplicates: {all_cap_df.shape[0]}")

# Remove duplicates based on 'Ticker' to ensure each company is listed only once
initial_companies = all_cap_df.shape[0]
all_cap_df.drop_duplicates(subset=['Ticker'], inplace=True)
final_companies = all_cap_df.shape[0]

print(f"Combined DataFrame shape after dropping duplicates: {all_cap_df.shape[0]}")
print(f"Removed {initial_companies - final_companies} duplicate entries.")

# Display the first few rows of the combined and deduplicated DataFrame
display(all_cap_df.head())
```
